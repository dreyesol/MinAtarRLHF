```python
import pandas as pd

import matplotlib.pyplot as plt

# Load the CSV file
file_path = 'data.csv'
data = pd.read_csv(file_path)

# Extract X and Y values
x_values = data.iloc[:, 0]
y1_values = data.iloc[:, 1]
y2_values = data.iloc[:, 2]

plt.figure(figsize=(10, 5))
plt.plot(x_values, y1_values, color='blue')
plt.xlabel('Training Episode')
plt.ylabel('Average Frames Survived')
plt.title('Average Frames Survived by Training Agent')
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(x_values, y2_values, color='green')
plt.xlabel('Training Episode')
plt.ylabel('Average G per Episode')
plt.title('Average G per Episode by Training Agent')
plt.legend()
plt.grid(True)
plt.show()
```
